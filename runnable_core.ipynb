```python
from abc import ABC,abstractmethod
```

```python
class Runnable(ABC):
    @abstractmethod
    def invoke(input_data):
      pass
```

```python
import random
class naklillm(Runnable):
    def __init__(self):
        print('llm created ')

    def invoke(self,prompt):
        response_list=[
            'delhi is the captal of india',
            'ipl is create league',
            'ai stands for artifical inteeligence'
        ]
        return {'response': random.choice(response_list)}
    def  predict(self,prompt):
        response_list=[
            'delhi is the captal of india',
            'ipl is create league',
            'ai stands for artifical inteeligence'
        ]
        return {'response': random.choice(response_list)}
```

```python
llm =naklillm()
print(llm.predict('what is ai'))
```

```python
class naklitemplete(Runnable) :
  def __init__(self,templete,input_variables):
    self.templete=templete
    self.input_variables=input_variables

  def invoke(self,input_dict):
    return self.templete.format(**input_dict)


  def format(self,input_dict):
    return self.templete.format(**input_dict)
```

```python
templete=naklitemplete(
    templete='what is the capital of {country}',
    input_variables=['country']
)
templete.format({'country':'india'})
```

```python
class naklillmchain:
  def __init__(self,llm,templete):
    self.llm=llm
    self.templete=templete

  def run(self,input_dict):
     final_prompt=self.templete.format(input_dict)
     return self.llm.predict(final_prompt)
```

```python
llm=naklillm()
templete=naklitemplete(
    templete='what is the capital of {country}',
    input_variables=['country'])
chain=naklillmchain(llm,templete)
```

```python
print(chain.run({'country':'india'}))
```

```python
class runnableconnector(Runnable):
  def __init__(self,runnable_list):
    self.runnable_list=runnable_list
  def invoke(self,input_data):
    for runnable in self.runnable_list :
      input_data=runnable.invoke(input_data)
    return input_data
```

```python
templete=naklitemplete(
    templete='what is the capital of {country}',
    input_variables=['country'])
llm=naklillm()
```

```python
chain=runnableconnector([templete, llm])
chain.invoke({'country':'india'})
```
